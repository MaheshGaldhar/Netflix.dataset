```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Fill missing numerical values with the median
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Fill missing text values with the mode
cat_cols = df.select_dtypes(include=["object", "category"]).columns

for col in cat_cols:
    if df[col].notna().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Clean text formats
for col in cat_cols:
    df[col] = df[col].astype("string[pyarrow]").str.strip().str.lower()

# Convert date columns
for col in df.columns:
    if "date" in col.lower():
        df[col] = pd.to_datetime(df[col], errors="coerce")

df.info()
```

```python
from pathlib import Path
clean_dataset=Path("cleaned_dataset.csv")
df.to_csv(clean_dataset,index=False)
print("cleaned dataset saved at:",clean_dataset.resolve())
```
