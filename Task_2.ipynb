```python
# import libraries 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# step 1
# Upload Celaned dataset into DataFrame
df=pd.read_csv("cleaned_dataset.csv")
```

```python
# Analyze dataset structure and statistics
df.head(5)
```

```python
# Dataset Struture
print("Rows and Columns:",df.shape)
```

```python
# dataset information
print("Dataset Information:",df.info())
```

```python
# Statistical information
print(df.describe(include="number"))
```

```python
# check all columns
print("Columns:",df.columns)
```

```python
# Checking Missing values 
print("MIssing Values from Dataset:",df.isnull().sum())
```

```python
# count distribution of type 
count_type=df["type"].value_counts()
print(count_type)
```

```python
# step 2
# Study content distribution by type
count_type.plot(kind="bar",figsize=(7,5),color=["red","Black"])
plt.xticks(rotation=0)
plt.title("Dataset distribution by Type")
plt.xlabel("Types")
plt.ylabel("Distribution value")
plt.legend([count_type])
plt.show()
```

```python
# step 3
#Identify top countries and categories.
country_data=(df["country"].dropna().str.split(",").explode().str.strip())
top_countrys=country_data.value_counts().head(10)
print(top_countrys)
```

```python
plt.figure(figsize=(10,6))
top_countrys.sort_values().plot(kind="barh")
plt.title(" Top 10 countrys by dataset content")
plt.ylabel("country")
plt.xlabel(" Number of Titles")
plt.show()
```

```python
# step 3
#Identify Top 10 categories/genres
genre_data=(df["listed_in"].dropna().str.split(",").explode().str.strip())
top_genre=genre_data.value_counts().head(10)
print(top_genre)
```

```python
plt.figure(figsize=(10,6))
top_genre.sort_values().plot(kind="barh")
plt.title(" Top 10 categories by dataset content")
plt.ylabel("categories")
plt.xlabel(" Number of genration")
plt.show()
```

```python
# step 4
year_counts = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(12,6))

plt.plot(year_counts.index, year_counts.values)

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
type_percentage=df["type"].value_counts(normalize=True)*100
print(type_percentage.round(2))
```

```python
# step 5
print("EDA SUMMARY")
print("1. Total titles:", len(df))

print("\n2. Content Type Distribution:")
print(df["type"].value_counts())

print("\n3. Top 10 Countries:")
print(top_countrys)

print("\n4. Top 10 Categories:")
print(top_genre)

print("\n5. Most common release year:")
print(df["release_year"].value_counts().idxmax())
```
